```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1x10 random numbers
numbers = np.random.rand(10)

print("Random Numbers:")
print(numbers)

# Plot histogram
plt.hist(numbers, bins=5, edgecolor='black')
plt.title("Uniform Random Numbers")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

2.Normally numbers

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 500 normal random numbers
numbers = np.random.normal(0, 1, 500)

print("First 10 Random Numbers:")
print(numbers[:10])

# Plot histogram
plt.hist(numbers, bins=20, edgecolor='black')
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

3.Random intiger with fixed seed

```python
import numpy as np

# Set seed
np.random.seed(10)

# Generate 3x3 matrix
matrix = np.random.randint(1, 51, size=(3, 3))

print("3x3 Matrix with Seed 10:")
print(matrix)
```

Without seed

4.T-test

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

data = np.array([5.2, 5.4, 5.1, 5.6, 5.3, 5.8])

# Hypothesized mean
mean = 5.5

t_statistic, p_value = stats.ttest_1samp(data, mean)

print("Data:", data)
print("Hypothesized Mean:", mean)
print("T-statistic:", t_statistic)
print("P-value:", p_value)

# Significance level
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")

# Visualization
plt.hist(data, bins=5, edgecolor='black')
plt.axvline(mean, linestyle='--', label='Mean = 5.5')

plt.title("One-Sample T-Test")
plt.xlabel("Data")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
