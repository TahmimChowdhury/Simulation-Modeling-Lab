```python
import numpy as np
from scipy import stats

data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])

alpha = 0.05

n1 = len(data1)
n2 = len(data2)

mean1 = np.mean(data1)
mean2 = np.mean(data2)

var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)

mean_diff = mean1 - mean2

pooled_var = (
    (n1 - 1) * var1 +
    (n2 - 1) * var2
) / (n1 + n2 - 2)

pooled_std = np.sqrt(pooled_var)

SE = pooled_std * np.sqrt(1/n1 + 1/n2)

t_stat = mean_diff / SE

df = n1 + n2 - 2

# Two-tailed p-value
p_value = 2 * stats.t.sf(abs(t_stat), df)

t_critical = stats.t.ppf(1 - alpha/2, df)

# 95% confidence interval
lower = mean_diff - t_critical * SE
upper = mean_diff + t_critical * SE

print("Mean 1:", mean1)
print("Mean 2:", mean2)
print("Variance 1:", var1)
print("Variance 2:", var2)
print("Mean Difference:", mean_diff)
print("Pooled Variance:", pooled_var)
print("Standard Error:", SE)
print("t-statistic:", t_stat)
print("Degrees of Freedom:", df)
print("p-value:", p_value)
print("95% CI:", (lower, upper))

if p_value < alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```
