```python



import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a 2x3 matrix with float values
matrix = np.array([[2.1, 4.7, 1.5],
                   [6.3, 3.8, 5.2]])

print("Matrix:")
print(matrix)


max_value = np.max(matrix)
min_value = np.min(matrix)

max_index = np.unravel_index(np.argmax(matrix), matrix.shape)
min_index = np.unravel_index(np.argmin(matrix), matrix.shape)

print("\nMaximum Value:", max_value)
print("Maximum Index:", max_index)

print("\nMinimum Value:", min_value)
print("Minimum Index:", min_index)


# Vector Length (Euclidean Norm)

vector_length = np.linalg.norm(matrix)
print("\nVector Length:", vector_length)


# Sorting

# Row-wise descending
row_desc = np.sort(matrix, axis=1)[:, ::-1]
print("\nRow-wise Descending Sort:")
print(row_desc)

# Entire matrix descending
entire_desc = np.sort(matrix.flatten())[::-1]
print("\nEntire Matrix Descending:")
print(entire_desc)


# Sum and Product

print("\nSum:", np.sum(matrix))
print("Product:", np.prod(matrix))


# Median, Mean, Standard Deviation

print("\nMedian:", np.median(matrix))
print("Mean:", np.mean(matrix))
print("Standard Deviation:", np.std(matrix))


# Bonus: Heatmap Visualization

plt.figure(figsize=(5, 3))
sns.heatmap(matrix, annot=True, cmap="YlGnBu", fmt=".1f")
plt.title("2×3 Matrix Heatmap")
plt.show()


# Bonus: Random Matrix Example

random_matrix = np.random.uniform(1, 10, (2, 3))

print("\nRandom 2×3 Matrix:")
print(random_matrix)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Create a 3×2 matrix

matrix = np.array([
    [1.5, -2.3],
    [3.8, 4.2],
    [-5.7, 0.9]
])

print("Original Matrix:")
print(matrix)

# 1. Trigonometric Functions

print("\nSin:")
print(np.sin(matrix))

print("\nCos:")
print(np.cos(matrix))

print("\nTan:")
print(np.tan(matrix))

# Clip values between -1 and 1
clipped_matrix = np.clip(matrix, -1, 1)

print("\nArcsin:")
print(np.arcsin(clipped_matrix))

print("\nArccos:")
print(np.arccos(clipped_matrix))

print("\nArctan:")
print(np.arctan(matrix))


# 2. Exponential and Logarithmic Functions


print("\nExponential:")
print(np.exp(matrix))

# Logarithm requires positive numbers
positive_matrix = np.abs(matrix) + 1

print("\nNatural Log:")
print(np.log(positive_matrix))

print("\nLog base 10:")
print(np.log10(positive_matrix))


# 3. Absolute Value and Square Root

print("\nAbsolute Value:")
print(np.abs(matrix))

# Square root needs non-negative values
sqrt_matrix = np.sqrt(np.abs(matrix))

print("\nSquare Root:")
print(sqrt_matrix)


# 4. Remainder when divided by 2

print("\nRemainder after division by 2:")
print(np.remainder(matrix, 2))


# 5. Rounding Functions

print("\nRound:")
print(np.round(matrix))

print("\nFloor:")
print(np.floor(matrix))

print("\nCeil:")
print(np.ceil(matrix))


# 6. Vector Functions

vector = np.array([1, 2, 3, 4, 5])

print("\nVector:")
print(vector)

print("\nSum:")
print(np.sum(vector))

print("\nMean:")
print(np.mean(vector))

print("\nMaximum:")
print(np.max(vector))

print("\nMinimum:")
print(np.min(vector))

print("\nStandard Deviation:")
print(np.std(vector))


# Bonus: Random Matrix

random_matrix = np.random.uniform(-5, 5, (3, 2))

print("\nRandom Matrix:")
print(random_matrix)

# Bonus: Heatmap Visualization

plt.figure(figsize=(6, 4))
sns.heatmap(
    matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("3×2 Matrix Heatmap")
plt.show()
```
